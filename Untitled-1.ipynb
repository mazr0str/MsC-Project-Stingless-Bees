```python
# Cell 0: Setup
import datetime
from pprint import pprint

# Optional: use pandas for nicer tables if available
try:
    import pandas as pd
    HAS_PANDAS = True
except Exception:
    HAS_PANDAS = False

# Configuration: set to True to use toy data (no manual prompts)
USE_TOY = True

# Toy dataset sizes (12 x 12)
N_LOC = 12
N_GEN = 12
Locations = [f"L{i}" for i in range(1, N_LOC+1)]
Meliponines = [f"M{i}" for i in range(1, N_GEN+1)]

def now_iso():
    return datetime.datetime.now().isoformat(sep=' ', timespec='seconds')

def print_table(title, rows, columns=None, max_rows=20):
    print(f"\n=== {title} ===")
    if HAS_PANDAS:
        df = pd.DataFrame(rows, index=columns) if columns and isinstance(rows, dict) else pd.DataFrame(rows)
        display(df.head(max_rows))
    else:
        # simple fallback
        if isinstance(rows, dict):
            for k,v in rows.items():
                print(k, v)
        else:
            for r in rows[:max_rows]:
                print(r)
```

```python
# Cell 1: Stage 1 - Bee Library (static traits)
bee_library = {}

if USE_TOY:
    # Toy static traits for M1..M12
    for i, M in enumerate(Meliponines, start=1):
        bee_library[M] = {
            "species_count": (i % 4) + 1,
            "head_length": round(1.5 + 0.1*i, 2),
            "max_body_length": round(4.5 + 0.2*i, 2),   # important: used for fc
            "hamuli": (i % 5) + 1,
            "pterostigma": round(0.5 + 0.05*i, 2)
        }
else:
    for M in Meliponines:
        bee_library[M] = {}
        bee_library[M]["species_count"] = int(input(f"species count for {M}: "))
        bee_library[M]["head_length"] = float(input(f"head length for {M}: "))
        bee_library[M]["max_body_length"] = float(input(f"max body length for {M}: "))
        bee_library[M]["hamuli"] = int(input(f"hamuli count for {M}: "))
        bee_library[M]["pterostigma"] = float(input(f"pterostigma width for {M}: "))

# Console output: Meliponine Library Table
if HAS_PANDAS:
    df_bee = pd.DataFrame.from_dict(bee_library, orient='index')
    df_bee.index.name = "Genus"
    display(df_bee)
else:
    print_table("Meliponine Library Table", bee_library)
```

```python
# Cell 2: Stage 2 & 3 - Prepare locations and collect headcounts
location_table = {}
headcount = {}

# Initialize location records
for L in Locations:
    location_table[L] = {
        "shf": None, "Rd": None, "Cd": None, "Fl": None,
        "fc": None, "Wa": None, "Wd": None
    }

# Collect headcounts (L x M)
if USE_TOY:
    # Toy headcounts: small integers
    import random
    random.seed(0)
    for L in Locations:
        headcount[L] = {}
        for M in Meliponines:
            headcount[L][M] = random.randint(0, 15)
else:
    for L in Locations:
        headcount[L] = {}
        for M in Meliponines:
            headcount[L][M] = int(input(f"How many {M} observed at {L}? "))

# Console output: Headcount Matrix (with TotalIndividuals)
headcount_table = []
for L in Locations:
    row = {"Location": L}
    total = 0
    for M in Meliponines:
        row[M] = headcount[L][M]
        total += headcount[L][M]
    row["TotalIndividuals"] = total
    headcount_table.append(row)

if HAS_PANDAS:
    df_head = pd.DataFrame(headcount_table).set_index("Location")
    display(df_head)
else:
    print_table("Headcount Matrix", headcount_table)
```

```python
# Cell 3: Stage 4 - compute fc[L] = sum_M headcount[L][M] * max_body_length[M]
derived_info = {L: {} for L in Locations}

def compute_fc_for_all_locations():
    for L in Locations:
        missing = []
        total = 0.0
        for M in Meliponines:
            mb = bee_library.get(M, {}).get("max_body_length")
            if mb is None:
                missing.append(M)
            else:
                total += headcount[L][M] * mb
        if missing:
            location_table[L]["fc"] = None
            derived_info[L]["fc"] = {"status": "incomplete", "missing_max_body_length": missing}
            print(f"[{now_iso()}] fc incomplete for {L}: missing max_body_length for {missing}")
        else:
            location_table[L]["fc"] = total
            derived_info[L]["fc"] = {"status": "ok", "derived_from": ["headcount", "max_body_length"], "last_computed": now_iso()}
            print(f"[{now_iso()}] fc computed for {L}: {total:.3f}")

compute_fc_for_all_locations()

# Console output: Computed fc Table
fc_table = [{"Location": L, "fc": location_table[L]["fc"], **derived_info[L].get("fc", {})} for L in Locations]
if HAS_PANDAS:
    display(pd.DataFrame(fc_table).set_index("Location"))
else:
    print_table("Computed fc Table", fc_table)
```
